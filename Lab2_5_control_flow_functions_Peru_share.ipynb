## 1. Operadores de Comparación y Lógicos

Antes de tomar decisiones, necesitamos entender cómo Python compara valores.

### 1.1 Operadores de comparación

Estos operadores comparan dos valores y devuelven `True` o `False`.

| Operador | Significado | Ejemplo |
|----------|-------------|--------|
| `==` | Igual a | `5 == 5` → True |
| `!=` | No igual a | `5 != 3` → True |
| `<` | Menor que | `3 < 5` → True |
| `>` | Mayor que | `5 > 3` → True |
| `<=` | Menor o igual | `5 <= 5` → True |
| `>=` | Mayor o igual | `5 >= 3` → True |

```python
# Comparando temperaturas - datos del SENAMHI en Iquitos
temperatura = 31.5

print(f"Temperatura: {temperatura}°C")
print(f"¿Está por encima de 0°C?     {temperatura > 0}")
print(f"¿Es exactamente 30°C?        {temperatura == 30}")
print(f"¿Es 30°C o más?              {temperatura >= 30}")
print(f"¿Supera el umbral de calor?  {temperatura > 35}")
```

```python
# Comparando cadenas (orden alfabético) - nombres de especies
especie = "Swietenia macrophylla"

print(especie == "Swietenia macrophylla")           # True - coincidencia exacta
print(especie == "swietenia macrophylla")           # False - sensible a mayúsculas
print(especie.lower() == "swietenia macrophylla")   # True - comparar en minúsculas
```

### 1.2 Operadores lógicos

Combina múltiples condiciones con `and`, `or`, `not`.

```python
# Condiciones climáticas en Madre de Dios
temp = 29
humedad = 85

# AND - ambas condiciones deben ser verdaderas
condiciones_tropicales = temp > 25 and humedad > 75
print(f"¿Condiciones tropicales típicas? {condiciones_tropicales}")

# OR - al menos una condición debe ser verdadera
alerta_extrema = temp > 38 or humedad < 20
print(f"¿Alerta climática extrema?       {alerta_extrema}")

# NOT - invierte el booleano
es_estacion_seca = humedad < 50
print(f"¿Es estación seca?               {es_estacion_seca}")
print(f"¿NO es estación seca?            {not es_estacion_seca}")
```

```python
# Ejemplo práctico: control de calidad de datos NDVI amazónicos
ndvi = 0.78

# El NDVI debe estar entre -1 y 1
es_valido = ndvi >= -1 and ndvi <= 1
print(f"NDVI = {ndvi}, ¿Válido? {es_valido}")

# Forma pythónica con comparación encadenada:
es_valido = -1 <= ndvi <= 1
print(f"Usando comparación encadenada: {es_valido}")

# ¿Corresponde a vegetación densa? (típico en bosque amazónico)
es_bosque_denso = 0.6 <= ndvi <= 1.0
print(f"¿Bosque denso (NDVI ≥ 0.6)?    {es_bosque_denso}")
```

## 2. Sentencias Condicionales

Las condicionales permiten que tu código tome decisiones y ejecute distintas acciones según las condiciones.

### 2.1 La sentencia `if`

```python
# Alerta de temperatura en una estación del SENAMHI en Piura
temperatura = 38.5

if temperatura > 35:
    print("⚠️  Alerta: Temperatura extrema detectada!")
    print("Verificar calibración del sensor y condiciones de El Niño.")

print("Continuando con el análisis...")  # Siempre se ejecuta
```

**Importante:** Python usa la **indentación** (4 espacios) para definir bloques de código. Todo lo indentado bajo el `if` forma parte de ese bloque.

### 2.2 La sentencia `if-else`

```python
# Clasificación de cobertura vegetal con NDVI
# Aplicación: monitoreo de deforestación en Loreto
ndvi = 0.72

if ndvi > 0.5:
    clase_vegetacion = "Vegetación densa"
else:
    clase_vegetacion = "Vegetación escasa o sin vegetación"

print(f"NDVI: {ndvi} → {clase_vegetacion}")
```

### 2.3 La cadena `if-elif-else`

Para múltiples condiciones, usa `elif` (abreviatura de "else if").

```python
# Clasificación de cobertura según NDVI - clave para alertas de deforestación en Perú
ndvi = 0.42

if ndvi < 0:
    categoria = "Agua / Nieve"
elif ndvi < 0.1:
    categoria = "Suelo desnudo / Roca"
elif ndvi < 0.3:
    categoria = "Vegetación escasa"
elif ndvi < 0.6:
    categoria = "Vegetación moderada"
else:
    categoria = "Vegetación densa (bosque)"

print(f"NDVI: {ndvi} → Categoría: {categoria}")
```

### 3.4 Construir listas con bucles

```python
# Convertir precipitaciones de mm a pulgadas
# Datos de precipitación mensual en Iquitos (mm) - fuente: SENAMHI
precip_mm = [312.0, 265.4, 320.8, 271.2, 198.5, 142.3, 131.7, 122.8, 191.4, 256.7, 290.1, 308.9]

# Comenzar con lista vacía, agregar cada resultado
precip_pulgadas = []

for p in precip_mm:
    pulgadas = p / 25.4
    precip_pulgadas.append(round(pulgadas, 2))

print(f"En mm:       {precip_mm}")
print(f"En pulgadas: {precip_pulgadas}")
```

### 3.6 El bucle `while`

Usa `while` cuando no sabes de antemano cuántas iteraciones necesitas.

```python
# Simulación del descenso del caudal del río Mantaro
# durante la estación seca (temporada de estiaje)
caudal_m3s = 180   # caudal inicial en m³/s
semanas = 0

while caudal_m3s > 80:
    caudal_m3s -= 15    # Disminuye ~15 m³/s por semana
    semanas += 1
    print(f"Semana {semanas}: Caudal = {caudal_m3s} m³/s")

print(f"\nEl caudal alcanzó el umbral de alerta (<80 m³/s) en {semanas} semanas")
```

### 3.7 Control de bucles: `break` y `continue`

```python
# break - detener el bucle al encontrar un dato inválido
registros_campo = ["Caoba", "Cedro", "DESCONOCIDA", "Tornillo", "Capirona"]

for especie in registros_campo:
    if especie == "DESCONOCIDA":
        print("⚠️  Especie no identificada - deteniendo procesamiento!")
        break
    print(f"Procesando: {especie}")
```

```python
# continue - saltar datos faltantes en registros del SENAMHI
# -9999 es el código estándar de dato faltante en muchos sistemas
precipitaciones = [45.2, -9999, 38.7, -9999, 62.1, 51.5]

datos_validos = []
for valor in precipitaciones:
    if valor == -9999:
        continue   # Saltar datos faltantes
    datos_validos.append(valor)

print(f"Original:      {precipitaciones}")
print(f"Solo válidos:  {datos_validos}")
print(f"Promedio (mm): {sum(datos_validos)/len(datos_validos):.1f}")
```

### 4.2 Funciones con parámetros

```python
# Función para convertir temperatura
# Útil al integrar datos de estaciones internacionales con datos del SENAMHI
def celsius_a_fahrenheit(celsius):
    """Convierte temperatura de Celsius a Fahrenheit."""
    fahrenheit = (celsius * 9/5) + 32
    return fahrenheit

# Usar la función con temperaturas peruanas
print(f"Iquitos:  28.3°C = {celsius_a_fahrenheit(28.3)} °F")
print(f"Cusco:    12.5°C = {celsius_a_fahrenheit(12.5):.3f} °F")
print(f"Puno:      9.8°C = {celsius_a_fahrenheit(9.8):.1f} °F")

# Usar en un bucle
for ciudad, temp_c in [("Lima", 18.5), ("Arequipa", 14.2), ("Piura", 29.1)]:
    print(f"{ciudad}: {temp_c}°C = {celsius_a_fahrenheit(temp_c):.1f} °F")
```

```python
# Función con documentación completa (buena práctica)
def calcular_area_basal(dap_cm):
    """Calcula el área basal de un árbol a partir del DAP.

    Args:
        dap_cm: Diámetro a la Altura del Pecho en centímetros

    Returns:
        Área basal en centímetros cuadrados
    """
    # import math
    radio = dap_cm / 2
    area = math.pi * radio ** 2
    return area

# Probar con una caoba de Madre de Dios
ba = calcular_area_basal(52.7)
print(f"Área basal de la caoba (DAP=52.7 cm): {ba:.2f} cm²")
```

### 4.3 Valores de parámetros por defecto

```python
def clasificar_ndvi(ndvi, umbral=0.5):
    """Clasifica la vegetación basándose en el valor NDVI.

    Args:
        ndvi: Valor del NDVI
        umbral: Valor a partir del cual la vegetación es 'densa' (por defecto 0.5)
    """
    if ndvi >= umbral:
        return "Vegetación densa"
    else:
        return "Vegetación escasa / sin vegetación"

# Con umbral por defecto (0.5)
print(clasificar_ndvi(0.72))    # → Vegetación densa
print(clasificar_ndvi(0.32))    # → Vegetación escasa

# Con umbral personalizado para zona de ecotono andino-amazónico
print(clasificar_ndvi(0.32, umbral=0.25))
```

### 4.4 Retornar múltiples valores

```python
def analizar_temperaturas(lista_temps):
    """Calcula estadísticas básicas para una lista de temperaturas.

    Returns:
        Tupla de (mínimo, máximo, promedio)
    """
    temp_min  = min(lista_temps)
    temp_max  = max(lista_temps)
    temp_prom = sum(lista_temps) / len(lista_temps)

    return temp_min, temp_max, temp_prom

# Temperaturas máximas en Lima durante el verano 2024
temps_lima_verano = [26.8, 27.5, 28.2, 27.1, 29.0, 26.3, 28.8]

minimo, maximo, promedio = analizar_temperaturas(temps_lima_verano)

print("Estadísticas de temperatura - Lima (verano 2024):")
print(f"  Mínima:   {minimo}°C")
print(f"  Máxima:   {maximo}°C")
print(f"  Promedio: {promedio:.1f}°C")
```

### 4.5 Ejemplo práctico: Calculadora de índices de vegetación (NDVI)

```python
def calcular_ndvi(nir, rojo):
    """Calcula el NDVI a partir de reflectancias NIR y Roja.

    NDVI = (NIR - Rojo) / (NIR + Rojo)

    Args:
        nir:  Reflectancia en el infrarrojo cercano
        rojo: Reflectancia en la banda roja

    Returns:
        Valor NDVI, o None si el cálculo no es válido
    """
    if nir + rojo == 0:
        return None  # Evitar división por cero

    ndvi = (nir - rojo) / (nir + rojo)
    return ndvi

# Datos de píxeles Sentinel-2 sobre distintos ecosistemas del Perú
datos_pixeles = [
    (0.52, 0.08, "Bosque amazónico maduro - Manu"),
    (0.25, 0.20, "Suelo desnudo - Pampas de La Joya, Arequipa"),
    (0.04, 0.12, "Superficie del Lago Titicaca"),
    (0.48, 0.10, "Agricultura en Chancay-Huaral"),
]

for nir, rojo, descripcion in datos_pixeles:
    ndvi = calcular_ndvi(nir, rojo)
    print(f"NIR={nir}, Rojo={rojo} → NDVI={ndvi:.3f}  [{descripcion}]")
```

### 4.6 Combinando funciones

```python
def clasificar_cobertura(ndvi):
    """Clasifica la cobertura del suelo según el valor NDVI."""
    if ndvi is None:
        return "Inválido"
    elif ndvi < 0:
        return "Cuerpo de agua"
    elif ndvi < 0.1:
        return "Suelo desnudo / Urbano"
    elif ndvi < 0.3:
        return "Vegetación escasa / Pastizal"
    elif ndvi < 0.6:
        return "Vegetación moderada / Agricultura"
    else:
        return "Bosque denso"


def procesar_pixel(nir, rojo):
    """Flujo completo: calcula NDVI y clasifica la cobertura."""
    ndvi = calcular_ndvi(nir, rojo)
    clasificacion = clasificar_cobertura(ndvi)
    return ndvi, clasificacion


# Análisis de píxeles de una imagen Landsat sobre el corredor
# Manu-Tambopata (Madre de Dios) (tuplas dentro de una lista)

#Inmutabilidad (Seguridad de los datos)-------------------------------------------------------------------
# Las tuplas son inmutables, lo que significa que una vez creadas, no se pueden cambiar.
# Tupla: Un píxel de una imagen satelital es una medición física fija. No quieres que, por error, una parte de tu código cambie el valor de rojo sin querer. La tupla protege esa integridad.
# Lista: Está diseñada para crecer, encogerse o cambiar.

pixeles = [
    (0.55, 0.07),   # Bosque primario
    (0.22, 0.19),   # Área deforestada reciente
    (0.04, 0.14),   # Río Madre de Dios
    (0.38, 0.12),   # Agricultura / chacras
]

print("Análisis de Cobertura - Corredor Manu-Tambopata")
print("-" * 55)

for i, (nir, rojo) in enumerate(pixeles, 1): # Enumerate da el indice y el valor desempaquetado al mismo tiempo de la lista
    ndvi, cobertura = procesar_pixel(nir, rojo)
    print(f"Píxel {i}: NDVI = {ndvi:.3f} → {cobertura}")
```

### 5.3 Análisis de datos CSV

```python
# Análisis básico de CSV
temperaturas = []

with open('temperaturas_senamhi.csv', 'r') as f:
    encabezado = f.readline()  # Saltar la primera línea

    for linea in f:
        partes = linea.strip().split(',')
        fecha    = partes[0]
        estacion = partes[1]
        region   = partes[2]
        temp     = float(partes[3])

        temperaturas.append(temp)
        print(f"{fecha} | {estacion} ({region}): {temp}°C")

print(f"\nRegistros procesados: {temperaturas}")
# Estadísticas
print(f"\nTemperatura promedio: {sum(temperaturas)/len(temperaturas):.1f}°C")
print(f"Mínima: {min(temperaturas)}°C | Máxima: {max(temperaturas)}°C")
```

## 6. Integrando todo: Mini Proyecto

Combinemos todo lo aprendido en un flujo de trabajo completo de procesamiento de datos forestales peruanos.

```python
# Flujo de trabajo completo: Procesar inventario forestal de Madre de Dios

import csv
import math

# ---- Funciones ----

def calcular_area_basal(dap_cm):
    """Calcula el área basal a partir del DAP."""
    return math.pi * (dap_cm / 2) ** 2


def clasificar_arbol(dap_cm):
    """Clasifica el árbol según su clase de tamaño (sistema SERFOR)."""
    if dap_cm < 10:
        return "Brinzal"
    elif dap_cm < 25:
        return "Latizal"
    elif dap_cm < 50:
        return "Fustal menor"
    else:
        return "Fustal mayor"


def es_aprovechable(dap_cm, dap_minimo=40):
    """Determina si un árbol supera el DAP mínimo de aprovechamiento."""
    return dap_cm >= dap_minimo


def procesar_inventario(archivo_entrada, archivo_salida):
    """Lee datos de campo, calcula campos adicionales y escribe resultados."""

    resultados = []

    # Leer archivo de entrada
    with open(archivo_entrada, 'r') as f:
        lector = csv.DictReader(f)

        for fila in lector:
            especie  = fila['especie']
            dap      = float(fila['dap_cm'])
            altura   = float(fila['altura_m'])
            parcela  = fila['parcela']

            # Calcular campos derivados
            area_basal    = calcular_area_basal(dap)
            clase_tamaño  = clasificar_arbol(dap)
            aprovechable  = es_aprovechable(dap)

            resultados.append({
                'especie':          especie,
                'parcela':          parcela,
                'dap_cm':           dap,
                'altura_m':         altura,
                'area_basal_cm2':   round(area_basal, 2),
                'clase_tamaño':     clase_tamaño,
                'aprovechable':     "Sí" if aprovechable else "No"
            })

    # Escribir archivo de salida
    with open(archivo_salida, 'w', newline='') as f:
        campos = ['especie', 'parcela', 'dap_cm', 'altura_m',
                  'area_basal_cm2', 'clase_tamaño', 'aprovechable']
        escritor = csv.DictWriter(f, fieldnames=campos)
        escritor.writeheader()
        escritor.writerows(resultados)

    return resultados


# ---- Ejecutar el flujo de trabajo ----

resultados = procesar_inventario('inventario_mdd.csv', 'inventario_mdd_analizado.csv')

# Resumen
print("Procesamiento completado!")
print(f"Total de árboles procesados: {len(resultados)}\n")

print("Resultados:")
print("-" * 75)
for arbol in resultados:
    print(f"{arbol['especie']:28} | DAP: {arbol['dap_cm']:5.1f} cm | "
          f"AB: {arbol['area_basal_cm2']:8.2f} cm² | "
          f"{arbol['clase_tamaño']:14} | Aprovechable: {arbol['aprovechable']}")
```
