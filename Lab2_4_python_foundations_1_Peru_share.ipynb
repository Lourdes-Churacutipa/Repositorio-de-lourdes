**Convenciones (recomendadas):**
- Usar nombres descriptivos: `altura_arbol` en lugar de `h`.

- Evitar letras sueltas, excepto en bucles.

### 3.4 Operadores aritméticos

````python
# Aritmética básica
a = 20
b = 6

print("Suma:", a + b)               # 26
print("Resta:", a - b)              # 14
print("Multiplicación:", a * b)     # 120
print("División:", a / b)           # 3.333... (siempre devuelve float)
print("División entera:", a // b)   # 3 (redondea hacia abajo)
print("Módulo:", a % b)             # 2 (resto)
print("Exponente:", a ** 2)         # 400 (a al cuadrado)
````

````python
# Ejemplo práctico: Calcular el área basal de una caoba (Swietenia macrophylla)
# en la Reserva de Biósfera del Manu, Madre de Dios
# Área basal = π * (DAP/2)²

import math  # Veremos más sobre importaciones más adelante

dap_cm = 52.7        # DAP de una caoba adulta en Madre de Dios
radio_cm = dap_cm / 2
area_basal_cm2 = math.pi * radio_cm ** 2

print(f"DAP: {dap_cm} cm")
print(f"Área basal: {area_basal_cm2:.2f} cm²")
````

````python
# Conversión de temperatura: Fahrenheit a Celsius
# Útil al trabajar con datos de estaciones meteorológicas internacionales
# Fórmula: C = (F - 32) * 5/9

temp_fahrenheit = 89.6     # Temperatura típica en Iquitos en verano
temp_celsius = (temp_fahrenheit - 32) * 5/9

print(f"{temp_fahrenheit}°F = {temp_celsius:.1f}°C")
````

````python
# Ejemplo andino: Calcular la presión atmosférica estimada según la altitud
# Fórmula simplificada: P = P0 * (1 - 0.0000226 * h)^5.256
# donde P0 = 1013.25 hPa (presión al nivel del mar) y h = altitud en metros

P0 = 1013.25           # Presión al nivel del mar (hPa)
altitud_cusco = 3400   # Altitud de Cusco en metros
altitud_puno  = 3827   # Altitud de Puno en metros

presion_cusco = P0 * (1 - 0.0000226 * altitud_cusco) ** 5.256
presion_puno  = P0 * (1 - 0.0000226 * altitud_puno)  ** 5.256

print(f"Presión en Cusco ({altitud_cusco} m): {presion_cusco:.1f} hPa")
print(f"Presión en Puno  ({altitud_puno} m): {presion_puno:.1f} hPa")
print(f"Diferencia: {presion_cusco - presion_puno:.1f} hPa")
````

### 4.3 Indexación y segmentación de cadenas

Las cadenas se **indexan** a partir de la posición 0. Se puede acceder a caracteres individuales o subcadenas.

```
String:  P  i  n  u  s
Index:   0  1  2  3  4
Neg:    -5 -4 -3 -2 -1
```

````python
especie = "Cedrela odorata"

# Indexación: acceso a caracteres individuales
print(f"Primer carácter: {especie[0]}")
print(f"Quinto carácter: {especie[4]}")
print(f"Último carácter: {especie[-1]}")
````

````python
# Segmentación: extraer subcadenas [inicio:fin]
# Nota: el índice final NO se incluye

especie = "Cedrela odorata"

genero  = especie[0:7]    # Caracteres 0-6: "Cedrela"
epiteto = especie[8:15]   # Caracteres 8-14: "odorata"

print(f"Género: {genero}")
print(f"Epíteto: {epiteto}")
````

````python
# Atajos de segmentación con códigos de estaciones del SENAMHI
codigo = "PE_AMA_IQUI_2024"

print(codigo[:2])     # País: "PE"
print(codigo[3:6])    # Región: "AMA" (Amazonas)
print(codigo[-4:])    # Año: "2024"
print(codigo[::2])    # Cada 2 caracteres
````

### 5.3 Modificación de listas

A diferencia de las cadenas de texto, las listas son **mutables**; se pueden modificar después de su creación.

````python
# Modificar un elemento
aves_manu = ["Guacamayo azul-amarillo", "Loro de cabeza azul", "Tucán grande"]
print(f"Antes: {aves_manu}")

aves_manu[2] = "Tucán de garganta blanca"   # Corrección del nombre
print(f"Después: {aves_manu}")
````

````python
# Agregar registros conforme avanza el trabajo de campo
registros_campo = ["Cedrela odorata", "Swietenia macrophylla"]

registros_campo.append("Virola surinamensis")           # Agregar al final
print(f"Después de agregar: {registros_campo}")

registros_campo.insert(1, "Cedrelinga cateniformis")    # Insertar en posición 1
print(f"Después de insertar: {registros_campo}")

registros_campo.extend(["Calycophyllum spruceanum", "Dipteryx micrantha"])  # Agregar varios
print(f"Después de extender: {registros_campo}")
````

````python
# Eliminando elementos (corrección de errores de campo)
especies = ["Caoba", "Cedro", "Tornillo", "Capirona", "Caoba"]

especies.remove("Tornillo")     # Eliminar por valor (primera aparición)
print(f"Después de eliminar: {especies}")

extraido = especies.pop()       # Eliminar y devolver el último elemento
print(f"Extraído: {extraido}")
print(f"Después de extraer: {especies}")

del especies[0]                 # Eliminar por índice
print(f"Después de del: {especies}")
````

### 5.4 Operaciones comunes con listas

````python
# Ordenación - precipitaciones mensuales en Madre de Dios (mm)
precipitacion_mdd = [312.4, 241.3, 405.7, 198.2, 389.6]

precipitacion_mdd.sort()                    # Ordenación in situ (ascendente)
print(f"Ascendente: {precipitacion_mdd}")

precipitacion_mdd.sort(reverse=True)        # Ordenación descendente
print(f"Descendente: {precipitacion_mdd}")
````

````python
# sorted() devuelve una nueva lista sin modificar la original.
original = [312.4, 241.3, 405.7, 198.2, 389.6]
copia_ordenada = sorted(original)

print(f"Original: {original}")
print(f"Copia ordenada: {copia_ordenada}")
````
